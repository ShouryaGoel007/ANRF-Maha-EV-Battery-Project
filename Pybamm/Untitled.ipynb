```python
import numpy as np
import matplotlib.pyplot as plt

# Load CSV (skip header row)
data = np.loadtxt(r"C:\Users\Shourya\AppData\Local\pybamm\pybamm\Cache\v1.0.0\US06.csv", delimiter=",", skiprows=1)

# Extract columns
time = data[:, 0]
current = data[:, 1]

# Plot
plt.figure(figsize=(10,5))
plt.plot(time, current)

plt.xlabel("Time (s)")
plt.ylabel("Current (A)")
plt.title("Drive Cycle: Current vs Time")

plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Load WLTC data
data = np.loadtxt(
    r"C:\Users\Shourya\AppData\Local\pybamm\pybamm\Cache\v1.0.0\WLTC.csv",
    delimiter=",",
    skiprows=1
)

# Extract columns
time = data[:, 0]
power = data[:, 1]

# Plot
plt.figure(figsize=(10,5))
plt.plot(time, power)

plt.xlabel("Time (s)")
plt.ylabel("Power (W)")
plt.title("WLTC Drive Cycle: Power vs Time")

plt.grid()
plt.show()
```
